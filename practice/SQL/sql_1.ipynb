#### filter columns from India 

```sql
select * from batch_1.practice.customers 
where country = "India";
```

```sql
-- #Customers who signup after march
 select * from batch_1.practice.customers
 where signup_date > "2022-03-01";
```

```sql
-- Count total customer 

select COUNT(*) as total_customers from batch_1.practice.customers;
```

```sql
-- Count customer per counrty

select COUNT(*) as total_cust_per_country, country from batch_1.practice.customers 
GROUP BY country;
```

```sql
--  Find latest customer
SELECT * FROM batch_1.practice.customers ORDER BY signup_date DESC LIMIT 1;
```
